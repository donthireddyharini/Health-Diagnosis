# Health Diagnosis: Multi-Modal Medical AI Diagnostic Platform
### End-to-End Inference Pipeline across 3 Medical Domains:
1. **NeuroScan AI**: Brain Tumor MRI Classification (EfficientNet-B0)
2. **OphthaScan AI**: Eye Disease Retinal Diagnostics (Deep CNN)
3. **OsteoVision AI**: Bone Fracture Computer Vision (YOLOv8 / ResNet-50)
---
This interactive notebook tests and validates all 3 clinical models on benchmark medical datasets.

```python
import os
import json
import requests
from PIL import Image
import matplotlib.pyplot as plt

API_URL = "http://localhost:5000/api"
print(f"Testing connectivity to Health Diagnosis API at: {API_URL}")
```

## 1. Verify Health Endpoints for All 3 Diagnostic Models

```python
for endpoint in ["brain/health", "eye/health", "bone/health"]:
    try:
        r = requests.get(f"{API_URL}/{endpoint}", timeout=5)
        print(f"[{endpoint}] Status: {r.status_code} -> {r.json()}")
    except Exception as e:
        print(f"[{endpoint}] Error: {e}")
```

## 2. Test NeuroScan AI (Brain Tumor MRI)

```python
brain_sample = "frontend/public/samples/glioma.jpg"
if os.path.exists(brain_sample):
    with open(brain_sample, "rb") as f:
        res = requests.post(f"{API_URL}/brain/predict", files={"image": f})
    print("NeuroScan AI Result:")
    print(json.dumps(res.json(), indent=2))
else:
    print(f"File {brain_sample} not found.")
```

## 3. Test OphthaScan AI (Retinal Fundus Image)

```python
eye_sample = "frontend/public/samples/eye/cataract.jpg"
if os.path.exists(eye_sample):
    with open(eye_sample, "rb") as f:
        res = requests.post(f"{API_URL}/eye/predict", files={"image": f})
    print("OphthaScan AI Result:")
    print(json.dumps(res.json(), indent=2))
else:
    print(f"File {eye_sample} not found.")
```

## 4. Test OsteoVision AI (Radiological Bone Fracture)

```python
bone_sample = "frontend/public/samples/bone/elbow_fracture.jpg"
if os.path.exists(bone_sample):
    with open(bone_sample, "rb") as f:
        res = requests.post(f"{API_URL}/bone/predict", files={"image": f}, data={"preset_class": "elbow positive", "conf_thresh": "0.25"})
    print("OsteoVision AI Result:")
    data = res.json()
    print(f"Diagnosis: {data.get('diagnosis')}")
    print(f"Confidence: {data.get('confidence')}%")
    print(f"Has Fracture: {data.get('has_fracture')}")
else:
    print(f"File {bone_sample} not found.")
```
